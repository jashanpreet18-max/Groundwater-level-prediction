# Groundwater Level Prediction & Exploratory Data Analysis

This notebook demonstrates exploratory data analysis (EDA), time-series decomposition, lag feature engineering, and baseline comparison across time-series forecasting models (Naive Baseline, ARIMA, Random Forest, Gradient Boosting) on Central Ground Water Board (CGWB) & DWLR groundwater observation datasets.

```python
import os
import math
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 11
```

## 1. Load and Inspect Dataset

```python
csv_path = '../data/raw/cgwb_dwlr_sample.csv'
df = pd.read_csv(csv_path)
df['Date'] = pd.to_datetime(df['Date'])
print('Dataset Shape:', df.shape)
print('Monitoring Stations:', df['Station'].unique())
df.head()
```

## 2. Summary Statistics

```python
print('Summary Statistics by Station:')
df.groupby('Station')['Water_Level'].agg(['count', 'min', 'mean', 'max', 'std']).round(2)
```

## 3. Historical Water Level Trends by Station

```python
plt.figure(figsize=(14, 7))
for station, group in df.groupby('Station'):
    plt.plot(group['Date'], group['Water_Level'], label=station, linewidth=2)

plt.xlabel('Date')
plt.ylabel('Groundwater Level (mbgl - meters below ground level)')
plt.title('Historical Groundwater Depletion Trends Across Stations')
plt.gca().invert_yaxis() # Invert so higher water table (lower mbgl) is at top
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## 4. Seasonal Cyclicity Analysis (Monthly Distribution)

```python
df['Month'] = df['Date'].dt.month_name()
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']

plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Month', y='Water_Level', order=month_order, palette='crest')
plt.title('Monthly Groundwater Level Variation (Pre vs Post Monsoon Cycles)')
plt.ylabel('Water Level (mbgl)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 5. Feature Engineering: Lag & Rolling Features

```python
# Filter to a single station for modeling demonstration
amritsar_df = df[df['Station'] == 'Amritsar Central'].sort_values('Date').reset_index(drop=True)

for lag in [1, 2, 3, 6, 12]:
    amritsar_df[f'lag_{lag}'] = amritsar_df['Water_Level'].shift(lag)

amritsar_df['rolling_mean_3'] = amritsar_df['Water_Level'].shift(1).rolling(3).mean()
amritsar_df.dropna().head()
```

## 6. Chronological Train-Test Split (No Data Leakage)

```python
clean_df = amritsar_df.dropna().reset_index(drop=True)
split_idx = int(len(clean_df) * 0.8)

train_data = clean_df.iloc[:split_idx]
test_data = clean_df.iloc[split_idx:]

print(f'Train samples: {len(train_data)}, Test samples: {len(test_data)}')
print('Train Date Range:', train_data['Date'].min().strftime('%Y-%m-%d'), 'to', train_data['Date'].max().strftime('%Y-%m-%d'))
print('Test Date Range:', test_data['Date'].min().strftime('%Y-%m-%d'), 'to', test_data['Date'].max().strftime('%Y-%m-%d'))
```
